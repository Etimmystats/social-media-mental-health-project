```python
school_records = {
    "Student_Name":["Etim","Emmanuel","Holiness","Desmond"],
    "Subjects":["Maths","Arts","History","Science"],
    "Scores":[70,56,82,74]
}
df_school = pd.DataFrame(school_records)
df_school.set_index('Student_Name')
```

```python
sport_records = {
    "Student_Name":["Etim","Emmanuel","Kay","Pelumi"],
    "Relay":["100M","200M","500M","1000M"],
    "Scores":[70,56,82,74]
}
df_sport = pd.DataFrame(sport_records)
df_sport.set_index('Student_Name')
```

```python
df_school.merge(df_sport,how="outer").set_index('Student_Name')
```

```python
pd.concat([df_school,df_sport],join="inner")
```
