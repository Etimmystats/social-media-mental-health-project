```python
import json
import os
from collections import defaultdict
from datetime import datetime

DATA_FILE = "budget_data.json"

def load_data():
    """Loads existing budget data from a JSON text file or initializes a new dictionary structure."""
    if os.path.exists(DATA_FILE):
        with open(DATA_FILE, 'r') as file:
            return json.load(file)
    return {"income": [], "expenses": []}

def save_data(data):
    """Persists the dictionary structure to a text file."""
    with open(DATA_FILE, 'w') as file:
        json.dump(data, file, indent=4)

def add_entry(data, entry_type):
    """Interactively logs a new income or expense entry."""
    try:
        amount = float(input(f"Enter {entry_type} amount: $"))
        category = input(f"Enter {entry_type} category/source: ").strip()
        date_str = input("Enter date (YYYY-MM-DD) or press Enter for today: ").strip()
        
        if not date_str:
            date_str = datetime.today().strftime('%Y-%m-%d')
        else:
            # Validate date string format
            datetime.strptime(date_str, '%Y-%m-%d')
            
        entry = {"amount": amount, "category": category, "date": date_str}
        data[entry_type].append(entry)
        save_data(data)
        print(f"\n✓ Successfully logged {entry_type} of ${amount:.2f}.")
        
    except ValueError:
        print("\n✗ Error: Please enter numerical values for amounts and correct date formats.")

def view_summary(data):
    """Generates a monthly summary of income, expenses, and top categories."""
    target_month = input("Enter month for summary (YYYY-MM) or press Enter for all-time: ").strip()
    
    total_income = 0.0
    total_expense = 0.0
    category_totals = defaultdict(float)
    
    # Calculate income
    for inc in data["income"]:
        if not target_month or inc["date"].startswith(target_month):
            total_income += inc["amount"]
            
    # Calculate expenses and group by category
    for exp in data["expenses"]:
        if not target_month or exp["date"].startswith(target_month):
            total_expense += exp["amount"]
            category_totals[exp["category"]] += exp["amount"]
            
    net_balance = total_income - total_expense
    
    # Render Output
    print("\n" + "="*40)
    print(f" BUDGET SUMMARY: {target_month if target_month else 'ALL-TIME'}")
    print("="*40)
    print(f"Total Income:   ${total_income:.2f}")
    print(f"Total Expenses: ${total_expense:.2f}")
    print("-" * 40)
    print(f"Net Balance:    ${net_balance:.2f}")
    
    if category_totals:
        print("\nTop 3 Spending Categories:")
        # Sort categories by total amount in descending order
        sorted_categories = sorted(category_totals.items(), key=lambda item: item[1], reverse=True)
        for i, (cat, amount) in enumerate(sorted_categories[:3], 1):
            print(f"  {i}. {cat}: ${amount:.2f}")
    else:
        print("\nNo expenses recorded for this period.")
    print("="*40 + "\n")

def main():
    data = load_data()
    
    while True:
        print("\n--- PERSONAL BUDGET TRACKER ---")
        print("1. Log Income")
        print("2. Log Expense")
        print("3. View Monthly Summary")
        print("4. Exit")
        
        choice = input("Select an option (1-4): ").strip()
        
        if choice == '1':
            add_entry(data, "income")
        elif choice == '2':
            add_entry(data, "expenses")
        elif choice == '3':
            view_summary(data)
        elif choice == '4':
            print("Data saved. Exiting tracker...")
            break
        else:
            print("Invalid option. Please choose a number between 1 and 4.")

if __name__ == "__main__":
    main()
```
